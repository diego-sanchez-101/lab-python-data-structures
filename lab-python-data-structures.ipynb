```python
#1. Defining the products list
products = ["t-shirt", "mug", "hat", "book", "keychain"]
```

```python
#2. Creating an empty dictionary
inventory = {}
```

```python
#3. Filling in the inventory database
for item in products:
    quantity = int(input(f"Enter the quantity of {item}"))
    inventory[item] = quantity
```

```python
print(inventory)
```

```python
#4.Creating the empty set customer_order
customer_orders = set()
```

```python
#5. Setting the custom order
print(f"Available products: {products}")
for i in range(3):
    order = input("Choose a product ")
    customer_orders.add(order)
```

```python
#6 Getting the order
print(customer_orders)
```

```python
#7. Calculating the total of the kind of products ordered
total_products_ordered = len(customer_orders)
print(total_products_ordered)
```

```python
#8. Percentage of products ordered (#part and the whole)
percentage_products_ordered = total_products_ordered / len(products) *100
print(percentage_products_ordered)
#The customer has ordered 60% of the store's lineup
```

```python
order_status = (total_products_ordered, percentage_products_ordered)
print(order_status)
```

```python
#9.Updating the inventory
for item in customer_orders:
    inventory[item] = inventory[item] - 1
print(inventory)
```

```python
for item, quantity in inventory.items():
    print(f"{item}, {quantity}")
```
